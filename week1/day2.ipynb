## First - let's talk about the Chat Completions API

1. The simplest way to call an LLM
2. It's called Chat Completions because it's saying: "here is a conversation, please predict what should come next"
3. The Chat Completions API was invented by OpenAI, but it's so popular that everybody uses it!

### We will start by calling Groq again - but don't worry, other-provider examples are coming too!


````python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('GROQ_API_KEY')

if not api_key:
    raise ValueError("GROQ_API_KEY was not found. Add it to your .env file.")
elif api_key.strip() != api_key:
    raise ValueError("GROQ_API_KEY has leading or trailing whitespace; remove it from your .env file.")
else:
    print("API key found and looks good so far!")
````

## Do you know what an Endpoint is?

If not, please review the Technical Foundations guide in the guides folder

And, here is an endpoint that might interest you...

Groq provides an OpenAI-compatible endpoint, so you can use the OpenAI Python client with Groq's base URL and your Groq API key. This sends requests to Groq, not to OpenAI.

````python
import requests

from openai import OpenAI

groq_openai = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=api_key,
)
````

````python
response = groq_openai.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[{"role": "user", "content": "Tell me a fun fact"}],
)

response
````

````python
response.choices[0].message.content
````

# What is the Groq package?

It's known as a Python Client Library.

It's nothing more than a wrapper around making this exact call to the http endpoint.

It just allows you to work with nice Python code instead of messing around with janky json objects.

But that's it. It's open-source and lightweight. Some people think it contains Groq model code - it doesn't!


````python
# Create Groq client

from groq import Groq
from openai import OpenAI
groq = Groq()

response = groq.chat.completions.create(model="openai/gpt-oss-120b", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content

````

## Use the OpenAI client with Groq

Groq provides an OpenAI-compatible endpoint. You can use the OpenAI Python client by setting Groq's endpoint URL and passing your Groq API key. Requests go to Groq, not OpenAI.

For example:

```python
groq = OpenAI(base_url="https://api.groq.com/openai/v1", api_key="gsk_...")
groq.chat.completions.create(...)
```

Add your key from https://console.groq.com/keys to the `.env` file as `GROQ_API_KEY=...`, then run the next two cells.

````python
GROQ_BASE_URL = "https://api.groq.com/openai/v1"

load_dotenv(override=True)

groq_api_key = os.getenv("GROQ_API_KEY")

if not groq_api_key:
    print("No API key was found - add GROQ_API_KEY to your .env file, save it, and then run this cell again.")
elif groq_api_key.strip() != groq_api_key:
    raise ValueError("GROQ_API_KEY has leading or trailing whitespace; remove it from your .env file.")
else:
    print("Groq API key found and looks good so far!")
````

````python
groq_openai = OpenAI(base_url=GROQ_BASE_URL, api_key=groq_api_key)

response = groq_openai.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[{"role": "user", "content": "Tell me a fun fact"}],
)

response.choices[0].message.content
````
