To demonstrate your familiarity with the Groq API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from groq import Groq
from openai import OpenAI
from dotenv import load_dotenv
```

```python
# constants

MODEL_GROQ = 'openai/gpt-oss-120b'
MODEL_LLAMA = 'llama3.2:latest'
```

```python
# set up environment
load_dotenv()
```

```python
# here is the question; type over this to ask something new

question = """
Please answer me the following question. In 2026, which is the best niche to be popular on youtube?
"""
```

```python
# Get Groq to answer, with streaming
response = Groq().chat.completions.create(
    model=MODEL_GROQ,
    messages=[{"role": "user", "content": f"{question}"}],
    )
print(response.choices[0].message.content)
```

```python
# Get Llama 3.2 to answer
response = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama").chat.completions.create(model=MODEL_LLAMA,messages=[{"role": "user", "content": f"{question}"}])
print(response.choices[0].message.content)
```
