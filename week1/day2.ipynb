## First - let's talk about the Chat Completions API

1. The simplest way to call an LLM
2. It's called Chat Completions because it's saying: "here is a conversation, please predict what should come next"
3. The Chat Completions API was invented by OpenAI, but it's so popular that everybody uses it!

Groq provides an OpenAI-compatible endpoint, so you can use the OpenAI Python client with Groq's base URL and your Groq API key. This sends requests to Groq, not to OpenAI.

# What is the Groq package?

It's known as a Python Client Library.

It's nothing more than a wrapper around making this exact call to the http endpoint.

It just allows you to work with nice Python code instead of messing around with janky json objects.

But that's it. It's open-source and lightweight. Some people think it contains Groq model code - it doesn't!


````python
# Create Groq client

from groq import Groq
from openai import OpenAI
groq = Groq()

response = groq.chat.completions.create(model="openai/gpt-oss-120b", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content

````

## And then this great thing happened:

OpenAI's Chat Completions API was so popular, that the other model providers created endpoints that are identical.

They are known as the "OpenAI Compatible Endpoints".

For example, google made one here: https://generativelanguage.googleapis.com/v1beta/openai/

And OpenAI decided to be kind: they said, hey, you can just use the same client library that we made for GPT. We'll allow you to specify a different endpoint URL and a different key, to use another provider.

So you can use:

```python
gemini = OpenAI(base_url="https://generativelanguage.googleapis.com/v1beta/openai/", api_key="AIz....")
gemini.chat.completions.create(...)
```

And to be clear - even though OpenAI is in the code, we're only using this lightweight python client library to call the endpoint - there's no OpenAI model involved here.

If you're confused, please review Guide 9 in the Guides folder!

And now let's try it!

## THIS IS OPTIONAL - but if you wish to try out Google Gemini, please visit:

https://aistudio.google.com/

And set up your API key at

https://aistudio.google.com/api-keys

And then add your key to the `.env` file, being sure to Save the .env file after you change it:

`GOOGLE_API_KEY=AIz...`


````python
import os
from dotenv import load_dotenv
from openai import OpenAI

GROQ_BASE_URL = "https://api.groq.com/openai/v1"

load_dotenv(override=True)

groq_api_key = os.getenv("GROQ_API_KEY")

if not groq_api_key:
    print("No API key was found - please add GROQ_API_KEY to your .env file and save it!")
elif not groq_api_key.startswith("gsk_"):
    print("An API key was found, but it doesn't start with 'gsk_'. Check if it's copied correctly.")
else:
    print("Groq API key found and looks good!")

# Initialize the OpenAI client pointing to Groq's endpoint
client = OpenAI(
    base_url=GROQ_BASE_URL,
    api_key=groq_api_key,
)

# Run a completion using a Groq model
response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {"role": "user", "content": "Tell me a fun fact"}
    ],
)

print(response.choices[0].message.content)
````
